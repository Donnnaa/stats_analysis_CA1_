```python
## Binomial Distribution Analysis

A company classified as bankrupt is considered a success (1), while a non-bankrupt company is considered a failure (0).
```

```python
from scipy.stats import binom
import numpy as np
import matplotlib.pyplot as plt

p = df['Bankrupt?'].mean()

print("Probability of Bankruptcy:", p)
```

```python
n = 8

p_exactly_2 = binom.pmf(2, n=n, p=p)

p_at_most_2 = binom.cdf(2, n=n, p=p)

p_more_than_2 = 1 - p_at_most_2

print("P(X = 2):", p_exactly_2)
print("P(X <= 2):", p_at_most_2)
print("P(X > 2):", p_more_than_2)
```

```python
x = np.arange(0, n+1)

pmf_values = binom.pmf(x, n=n, p=p)

plt.figure(figsize=(8,5))
plt.bar(x, pmf_values)

plt.title("Binomial PMF")
plt.xlabel("Number of Bankrupt Companies")
plt.ylabel("Probability")

plt.show()
```

```python
cdf_values = binom.cdf(x, n=n, p=p)

plt.figure(figsize=(8,5))

plt.step(x, cdf_values, where='post',  marker='o')

plt.title("Binomial CDF")
plt.xlabel("Number of Bankrupt Companies")
plt.ylabel("Cumulative Probability")

plt.show()
```

```python
PMF shows the probability of observing exactly x bankrupt companies.

CDF shows the probability of observing x or fewer bankrupt companies.

The Binomial distribution is appropriate because bankruptcy has two possible outcomes: bankrupt or non-bankrupt.
```
