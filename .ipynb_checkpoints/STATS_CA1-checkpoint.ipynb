```python
# Binomial Distribution Analysis
```

```python
# Code for Binomial Distribution

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Probability of bankruptcy from dataset

p = df['Bankrupt?'].mean()

# Parameters

n = 10   # Number of companies observed

k = np.arange(0, n + 1)   # Possible bankruptcies (0 to 10)

# Binomial PMF

binom_pmf = binom.pmf(k, n, p)

# Plot the Binomial Distribution

plt.bar(k, binom_pmf, color='blue', edgecolor='black')

plt.title('Binomial Distribution of Bankruptcy Events')
plt.xlabel('Number of Bankrupt Companies')
plt.ylabel('Probability')

plt.show()
```

```python
# Probability of exactly 2 bankrupt companies

p_exactly_2 = binom.pmf(2, n, p)

print("P(X = 2):", p_exactly_2)
```

```python
# Probability of at most 2 bankrupt companies

p_at_most_2 = binom.cdf(2, n, p)

print("P(X <= 2):", p_at_most_2)
```

```python
# Probability of more than 2 bankrupt companies

p_more_than_2 = 1 - binom.cdf(2, n, p)

print("P(X > 2):", p_more_than_2)
```

```python
A Binomial distribution was used because the bankruptcy variable has only two possible outcomes: bankrupt (1) and non-bankrupt (0). 
    Each company can therefore be treated as a Bernoulli trial. 
        The graph shows the probability distribution of the number of bankrupt companies among 10 observations.
```

```python
## Poisson Distribution Analysis
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Parameters

lambda_bankruptcy = df['Bankrupt?'].mean() * 100

k = np.arange(0, 15)

# Poisson probability mass function (PMF)

poisson_pmf = poisson.pmf(k, lambda_bankruptcy)

# Plot the Poisson Distribution

plt.bar(k, poisson_pmf, color='pink', edgecolor='black')

plt.title(f'Poisson Distribution: λ = {lambda_bankruptcy:.2f}')
plt.xlabel('Number of Bankruptcy Events')
plt.ylabel('Probability')

plt.xticks(k)

plt.show()
```

```python
p_exactly_2 = poisson.pmf(2, lambda_bankruptcy)

print("P(X = 2):", p_exactly_2)
```

```python
# Probability of exactly 2 bankruptcy events

p_exactly_2 = poisson.pmf(2, lambda_bankruptcy)

print("P(X = 2):", p_exactly_2)
```

```python
# Probability of at most 2 bankruptcy events

p_at_most_2 = poisson.cdf(2, lambda_bankruptcy)

print("P(X <= 2):", p_at_most_2)
```

```python
The Poisson distribution was used to model the frequency of bankruptcy events. 
The parameter λ was estimated from the observed bankruptcy rate in the dataset. 
This distribution helps estimate the probability of a certain number of bankruptcy events occurring within a fixed sample.
```

```python
## Normal Distribution Analysis of ROA
```

```python
# Select ROA variable
 
roa = df[' ROA(A) before interest and % after tax']
 
# Calculate statistics
 
mean_roa = roa.mean()
std_roa = roa.std()
 
print("Mean:", mean_roa)
print("Standard Deviation:", std_roa)
print("Skewness:", roa.skew())
```

```python
# Code for Normal Distribution

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Generate x values

x = np.linspace(roa.min(), roa.max(), 100)

# Normal probability density function

y = norm.pdf(x, mean_roa, std_roa)

# Plot histogram

plt.hist(roa, bins=30, density=True, color='pink', edgecolor='black')

# Plot normal curve

plt.plot(x, y, color='blue', linewidth=2)

plt.title('Normal Distribution of ROA')
plt.xlabel('ROA')
plt.ylabel('Density')

plt.show()
```

```python
The ROA (Return on Assets) variable was selected because it is a continuous numerical variable.
The histogram shows the distribution of ROA values, while the red curve represents the theoretical normal distribution based on the calculated mean and standard deviation.
The skewness value was used to determine whether the data is symmetric or skewed. If the histogram follows the red curve closely, the data can be considered approximately normally distributed.
```
